```python
from ase.build import graphene_nanoribbon
atoms = graphene_nanoribbon(
    n=8,
    m=2,
    type='zigzag',
    saturated=True,
    C_C=1.42,
    C_H=1.09,
    vacuum=10.0,
    sheet=False
)
```

```python
from ase.visualize import view
view(atoms)
```

```python
import numpy as np
magmoms = []
for atom in atoms:
    #if atom.symbol == 'C':
    magmoms.append(np.random.uniform(-1, 1))

atoms.set_initial_magnetic_moments(magmoms)
```

```python
view(atoms)
```

```python
from gpaw.lcao.basis import Basis
print(Basis.get_basis_names())
```

```python
from gpaw import GPAW, FermiDirac
from gpaw.mpi import world
from gpaw import GPAW, FermiDirac

calc = GPAW(
    mode='lcao',             # Fast for relaxation; switch to 'pw' for final energies
    basis='dzp',             # Good compromise for C systems
    xc='PBE',                # Standard GGA
    spinpol=True,            # ❤️ Enable spin-polarisation
    occupations=FermiDirac(0.10),  # Smearing (small)
    txt='gnr_relax.txt',
    kpts=(1, 1, 1),         # period along z; adjust depending on orientation
)
atoms.calc = calc
```

```python
from ase.optimize import BFGS

dyn = BFGS(atoms, trajectory='gnr_relax.traj')
dyn.run(fmax=0.02)
```

```python
from ase.io import write, read
write('gnr_relax.xyz', atoms)
```
